#### ***Chaining***

##### **Chaining is sequence of steps means we can process the output of one act as input ofanother**

```python
###Load API Key
import os
from dotenv import load_dotenv
load_dotenv()

if os.environ['GROQ_API_KEY']:
    print("API Key is Set")
```

```python
### Create LLM
from langchain_groq import ChatGroq

llm = ChatGroq(
    model = "openai/gpt-oss-120b"
)

llm
```

```python
### Design a single text prompt using Prompt Template
from langchain_core.prompts import PromptTemplate

prompt_template_1 = PromptTemplate.from_template("""
Can you please write a detailed report about {topic}
""")
```

```python
prompt_template_2 = PromptTemplate.from_template("""
    Generate a 4 points summary about {text}
""")
```

```python
from langchain_core.output_parsers import StrOutputParser
chain = prompt_template_1 | llm | StrOutputParser() | prompt_template_2 | llm
```

```python
response = chain.invoke({
    "topic":"RRR"
})
```

```python
print(response.content)
```
