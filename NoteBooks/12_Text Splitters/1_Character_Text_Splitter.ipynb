### ***Text Splitter***

#### ***Text Splitter takes larger documents that divides into a smaller chunks that easier to embed,retrieve, and send to LLM***

#### ***1.Character Text Splitter***

##### ***Character Text Splitter is a Text Splitter is used to split the documents based on separator that creates chunks by using chunk_size.***

```python
from langchain_text_splitters import CharacterTextSplitter

text = """LangSmith is an observability and evaluation platform for LLM
applications. I use it to trace and debug LangChain, LangGraph, RAG,
and agent workflows; inspect prompts, model calls, retrieval results,
and tool executions; build datasets and evaluators; compare prompts
and models; perform regression testing; and monitor production
behavior using traces, feedback, latency, errors, and quality
metrics.""" 

splitter = CharacterTextSplitter(
    chunk_size = 300,
    chunk_overlap = 50,
    separator="\n\n"
)

chunks = splitter.split_text(text)

chunks
```

```python
for chunk in chunks:
    print(chunk)
    print(len(chunk))
    print("#"*40)
```

```python
### Lets split the document
import os
from langchain_community.document_loaders import TextLoader

path = "../Document Loaders/Docs/langsmith.txt"

if os.path.exists(path):
    print("Path is Exist")
else:
    print("Path is not Exist")

loader = TextLoader(path)

documents = loader.load()
```

```python
doc_chunks = splitter.split_documents(documents)

```

```python
for chunk in doc_chunks:
    print(chunk.page_content)
    print("-"*50)
```
