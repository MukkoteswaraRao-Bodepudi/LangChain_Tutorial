#### ***Python Code Text Splitter***
##### ***Python Code TextSplitter is a Text Splitter splits the Python code into a meaningful code chunks***

```python
from langchain_text_splitters import PythonCodeTextSplitter

splitter = PythonCodeTextSplitter(
    chunk_size = 100,
    chunk_overlap = 20
)
```

```python
code_text = """class Calculator:

    def add(self, a, b):
        return a + b

    def subtract(self, a, b):
        return a - b


class User:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_details(self):
        return f"{self.name} is {self.age} years old"


def calculate_average(numbers):
    total = sum(numbers)
    return total / len(numbers)


def main():
    calculator = Calculator()

    result = calculator.add(10, 20)
    print("Addition:", result)

    average = calculate_average([10, 20, 30])
    print("Average:", average)


if __name__ == "__main__":
    main()"""
```

```python
chunks = splitter.split_text(code_text)
chunks
```

```python
for chunk in chunks:
    print(chunk)
    print("#"*50)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter,Language

splitter1 = RecursiveCharacterTextSplitter.from_language(
    Language.PYTHON,
    chunk_size = 100,
    chunk_overlap=20
)
```

```python
chunks1 = splitter.split_text(code_text)
for chunk in chunks1:
    print(chunk)
    print("#"*50)
```
