#### ***Prompt Template***
##### ***Prompt Template is a reusable prompts where you can fill values dynamically.***
##### **Prompt Template is construct for single text prompt.**

```python
###Load API Key
import os
from dotenv import load_dotenv
load_dotenv()

if os.environ['GROQ_API_KEY']:
    print("API Key is Set")
```

```python
### Create LLM
from langchain_groq import ChatGroq

llm = ChatGroq(
    model = "openai/gpt-oss-120b"
)

llm
```

```python
### Design a single text prompt using Prompt Template
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate.from_template("""
Can you please write a detailed report about {topic}
""")
```

```python
prompt_1 = prompt_template.invoke({
    "topic":"Indian Priemer League"
})
prompt_1
```

```python
response = llm.invoke(prompt_1)
result_1=response.content
print(result_1)
```

```python
type(result_1)
```

```python
prompt_template = PromptTemplate.from_template("""
    Generate a 4 points summary about {text}
""")

prompt_2 = prompt_template.invoke({
    "text":result_1
})
prompt_2
```

```python
response = llm.invoke(prompt_2)
result_2 = response.content
print(result_2)
```
