#### ***Sequential Chains***

##### **Sequential Chains means run multiple chains one after another,where the output of one step becomes, the input of the next step.**

##### ***Load API Key***

```python
import os
from dotenv import load_dotenv
load_dotenv()

if os.environ['GROQ_API_KEY']:
    print("API Key is Set.")
else:
    print("API Key is not Set.")
```

##### ***LLM***

```python
### Create LLM
from langchain_groq import ChatGroq

llm = ChatGroq(
    model = "openai/gpt-oss-120b"
)

llm
```

#### ***Prompt Template 1***

```python
from langchain_core.prompts import ChatPromptTemplate
prompt_template_1 = ChatPromptTemplate.from_messages(
    [
        ("system","You are an Helpful AI Expert."),
        ("human","Can you please generate a detailed report about {topic}")
    ]
)
```

```python
from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
```

#### ***Prompt Template 2***

```python
from langchain_core.prompts import ChatPromptTemplate
prompt_template_2 = ChatPromptTemplate.from_messages(
    [
        ("system","You are an Helpful AI Expert."),
        ("human","Generate the four points summary on {text}")
    ]
)
```

##### ***Sequential Chain***

```python
chain_1 = prompt_template_1 | llm | output_parser
chain_2=prompt_template_2 | llm | output_parser

sequential_chain = chain_1 | chain_2

response = sequential_chain.invoke({"topic":"Harly Davidson"})

print(response)
```

```python
from langchain_core.runnables import RunnableSequence

final_chain = RunnableSequence(chain_1,chain_2)

response = final_chain.invoke({"topic":"Germany"})
print(response)
```
