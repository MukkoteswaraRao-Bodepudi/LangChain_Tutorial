#### ***Parallel Chains***

##### **Parallel Chains means run multiple chains each chain execute independently based on the same input rather than waiting for one after another.**

```python
import os
from dotenv import load_dotenv
load_dotenv()

if os.environ['GROQ_API_KEY']:
    print("API Key is Set.")
else:
    print("API Key is not Set.")
```

```python
### Create LLM
from langchain_groq import ChatGroq

llm = ChatGroq(
    model = "openai/gpt-oss-120b"
)

llm
```

```python
#### Design a promp
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    ("system","You are an Helpful AI Assistant."),
    ("human","Generate a meaningful definition without any examples about {topic}")
])
```

```python
#### Design a promp
from langchain_core.prompts import ChatPromptTemplate

interview_prompt = ChatPromptTemplate.from_messages([
    ("system","You are an Helpful AI Assistant."),
    ("human","Generate a 5 Interview questions and answer on {topic},without extra text.")
])
```

```python
#### Design a promp
from langchain_core.prompts import ChatPromptTemplate

coding_prompt = ChatPromptTemplate.from_messages([
    ("system","You are an Helpful AI Assistant."),
    ("human","Generate a code based on {topic},without extra text.")
])
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
```

```python
content_chain = prompt | llm | parser
coding_chain = coding_prompt | llm | parser
interview_chain = interview_prompt | llm | parser
```

```python
from langchain_core.runnables import RunnableParallel

parallel_chain = RunnableParallel(
    content = content_chain,
    coding = coding_chain,
    interview = interview_chain
)
```

```python
response = parallel_chain.invoke({"topic":"Functions in Python"})
print(response)
```

```python
for key,value in response.items():
    print(key,":",value)
    print("#"*50)
```
