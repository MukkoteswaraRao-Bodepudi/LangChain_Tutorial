#### ***Runnable Sequence***

##### **Runnable sequence is a LangChain Runnable that executes multiple Runnables sequentially,the output of each step to next step.**

```python
#### Load API Key

from dotenv import load_dotenv

load_dotenv()
```

```python

### Create LLM
from langchain_groq import ChatGroq

llm = ChatGroq(
    model = "openai/gpt-oss-120b"
)

llm
```

```python
### Prompt 1
from langchain_core.prompts import PromptTemplate
prompt_template_1 = PromptTemplate.from_template("""
        Generate a Joke on {topic}, one short line.
""")
```

```python
### Prompt 12
prompt_template_2 = PromptTemplate.from_template("""
        Explain a Joke from {text}.

        if it's possible display the Text:{text},

        Explanation:
""")
```

```python
###output parser
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
```

```python
from langchain_core.runnables import RunnableSequence

chain = RunnableSequence(prompt_template_1,llm,parser,prompt_template_2,llm,parser)

response = chain.invoke({"topic":"Human"})
print(response)
```
