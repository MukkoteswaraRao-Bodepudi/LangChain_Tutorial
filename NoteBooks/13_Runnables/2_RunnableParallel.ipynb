#### **Runnable Parallel**

##### **Runnable Parallel is a LangChain Runnable that executes multiple runnables at the same time and based on the same input and collect their results in a dictionary.**

```python
#### Load API Key

from dotenv import load_dotenv

load_dotenv()
```

```python

### Create LLM
from langchain_groq import ChatGroq

llm = ChatGroq(
    model = "openai/gpt-oss-120b"
)

llm
```

```python
### Prompt 1
from langchain_core.prompts import PromptTemplate

linked_prompt = PromptTemplate.from_template("""
    You are an Linked content Expert,

    Generate a linkedin post based on {topic}, without extra text.
""")
```

```python
### Prompt 2
tweet_prompt = PromptTemplate.from_template("""
    You are an Tweet content Expert,

    Generate a tweet post based on {topic}, without extra text.
""")
```

```python
from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()
```

```python
from langchain_core.runnables import RunnableSequence,RunnableParallel

chain = RunnableParallel({
    "LinkedIn":RunnableSequence(linked_prompt,llm,output_parser),
    "Tweet":RunnableSequence(tweet_prompt,llm,output_parser)
})

responses = chain.invoke({"topic":"LangSmith"})

for key,value in responses.items():
    print(key+":"+value)
    print("#"*100)
```
