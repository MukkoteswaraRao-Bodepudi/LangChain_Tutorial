### ***Retriever in LangChain***

#### **Retriever is a Component in a Rag System that takes user's query and find's the relevant chunks from the knowledge base.**

```python
### Load the enivronment variable
from dotenv import load_dotenv
load_dotenv()
```

```python
###Create a same embedding model which was create a embedding vector for chunks

### Create a Embedding model by HuggingFace with 1024 dimensions
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
        model = "BAAI/bge-large-en-v1.5",
        model_kwargs = {"device":"cpu"},
        encode_kwargs = {"normalize_embeddings":True}
)
```

```python
#### we have already chroma persist directory just load it

from langchain_chroma import Chroma
vectorstore = Chroma(
    embedding_function=embedding_model,
    persist_directory="../vectorstore/rishabh_pant",
    collection_name="rishabh_pant"
)
```

```python
### create a retriever using vector store for similarity search

retriever = vectorstore.as_retriever(search_kwargs={"k":3})
```

```python
### User Query 
user_query = "When did Rishabh Pant make his T20I, Test, and ODI debuts, and what major record does he hold in Test cricket?"
```

```python
results = retriever.invoke(user_query)
print(results)
```

```python
for doc in results:
    print("Content:",doc.page_content)
```
