##### ***Embeddings***
##### Embeddings are numerical representation of Text

#### ***Create a Embedding Model using Huggingface***

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
        model = "BAAI/bge-large-en-v1.5",
        model_kwargs = {"device":"cpu"},
        encode_kwargs = {"normalize_embeddings":True}
)
```

```python
### create a Embedding Vectore for query

query = "Delhi is the Capital of India"

query_vector = embedding_model.embed_query(query)

print(len(query_vector))  #Dimensions
```

```python
#### Create a Embedding Vector for Document

documents = [
    "Assembely is located in Delhi",
    "Delhi is high polluted city in India",
    "Paris is the Capital of France"
]

document_vector = embedding_model.embed_documents(documents)
print(len(document_vector)) ### Number of vectors
```

```python
from langchain_core.vectorstores.in_memory import cosine_similarity
#### Cosine similarity between query_vector and document vectors

similarity_score = cosine_similarity([query_vector],document_vector)
print(similarity_score)
```

```python
### Another Example
query1 = "tea"
documents_2 = ["milk","coffe","apple"]

query_vector_1 = embedding_model.embed_query(query1)

document_vector_1 = embedding_model.embed_documents(documents_2)

score = cosine_similarity([query_vector_1],document_vector_1)
print(score)
```
