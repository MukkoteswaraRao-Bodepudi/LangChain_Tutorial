### ***Tools in LangChain***

#### ***A Tool is a function in LangChain that an LLM can choose to call  when it's need to perform an actions***

```python
from langchain_core.tools import tool

@tool
def multiply(a:int,b:int)->int:
    """Multiply Two Numbers"""

    return a*b
```

```python
result = multiply.invoke({"a":10,"b":5})
print(result)
```
