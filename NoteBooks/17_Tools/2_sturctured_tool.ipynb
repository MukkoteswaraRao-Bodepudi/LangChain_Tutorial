#### ***Structured Tool in LangChain***
##### ***Structured Tool is a LangChain that has defined input Schema***

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel,Field
```

```python
### Pydantic Object

class Multiply(BaseModel):
    a:int = Field(description = "The first value to multiply")
    b:int = Field(description = "The Second Value to multiply")

    
```

```python
## function
def multiply_func(a:int,b:int):
    """Multiply two numbers"""

    return a*b
```

```python
tool = StructuredTool.from_function(
    func=multiply_func,
    name = "Multipy Function",
    description="The Function is used to multiply Two numbers",
    args_schema=Multiply
)
```

```python
result = tool.invoke({"a":20,"b":7})
print(result)
```
