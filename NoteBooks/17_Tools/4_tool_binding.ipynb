#### ***Tool Binding***

##### **Tool Binding is the process of connecting tools to an LLM,so the model knows which tools are available and can request their use.**

```python
from langchain_core.tools import tool
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv()
```

```python
llm = ChatGroq(
    model = "openai/gpt-oss-120b"
)
llm
```

```python
@tool
def add(a:int,b:int)->int:

    """Add Two Numbers"""
    return a+b
```

```python
tools = [add]

llm_with_tools = llm.bind_tools(tools)

result = llm_with_tools.invoke("can you add 5 with 19")
print(result)
```
