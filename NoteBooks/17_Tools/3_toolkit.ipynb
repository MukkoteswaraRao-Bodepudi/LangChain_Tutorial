#### ***ToolKit***

##### **ToolKit is a collection of tools packed together for specific purpose.**

```python
from langchain_core.tools import tool
```

```python
##tool1 
@tool
def add(a:int,b:int)->int:
    "Add Two Numbersr"
    return a+b
```

```python
## Tool2
@tool
def multi(a:int,b:int)->int:
    """Multiply Two Numbers"""
    return a*b
```

```python
#tool3
@tool
def subt(a:int,b:int)->int:
    """Subtract Two Numbers"""
    return (a-b)
```

```python
## Tool Kit using Python Class
class Mathkit():

    def get_tools(self):
        return [add,multi,subt]
```

```python
### Create a Toolkit Object for Mathkit class

toolkit = Mathkit()

tools = toolkit.get_tools()

for tool in tools:
    print(tool.name+" --> "+tool.description)
```
