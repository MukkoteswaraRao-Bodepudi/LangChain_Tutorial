### **Rishabh Pant AI Knowledge Assistant by Wikipedia Content Only**

### ***Vector Store in LangChain***

#### ***Vector Store is a database is design to store vector embeddings and efficiently find the vector's that are similar to a user's query.***

```python
### Load Environment variables
from dotenv import load_dotenv
load_dotenv()
```

```python
###Load the document from webpage url

from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader("https://en.wikipedia.org/wiki/Rishabh_Pant")

documents = loader.load()

print("Number Of Documents:",len(documents))
```

```python
documents[0].metadata
```

```python
documents[0].page_content[:500]
```

```python
### Initialize the Splitter using RecursiveCharacterTextSplitter

from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size = 700,
    chunk_overlap = 100
)
```

```python
### Create a chunks by using chunk size
chunks = splitter.split_documents(documents)
print("Number Of Chunks:",len(chunks))
```

```python
### Create a Embedding model by HuggingFace with 1024 dimensions
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
        model = "BAAI/bge-large-en-v1.5",
        model_kwargs = {"device":"cpu"},
        encode_kwargs = {"normalize_embeddings":True}
)
```

```python
### Create and Store vector embedding using Chroma

from langchain_chroma import Chroma

vectorstore = Chroma.from_documents(
    documents = chunks,
    embedding=embedding_model,
    persist_directory="../vectorstore/rishabh_pant",
    collection_name="rishabh_pant"
)
```

```python
### User query
user_query = "Who is Rishabh Pant, and what are the major milestones and achievements of his cricket career?"

```

```python
#### Similarity search means vectorstore find the relevant documents,
#the embedding vector which are close to the user's query.

retrieved_documents = vectorstore.similarity_search(
    user_query,
    k=3
)
```

```python
user_query = "When did Rishabh Pant make his T20I, Test, and ODI debuts?"
```

```python
## Similarity search with score , less score means more similar
retrieved_documents = vectorstore.similarity_search_with_score(
    user_query,
    k=3
)
```

```python
print(retrieved_documents)
```
