```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model = "openai/gpt-oss-120b"
)

llm
```

#### ***Messages in LangChain***

##### **Messages are structure of conversation data that contains role and content**

```python
# Without Message
response = llm.invoke("What is LangChain?")
print(response.content)
```

#### **Types Of Messages**

#### ***1.System Message***
##### **System Messages are used to control the behaviour of LLM/AI**

#### ***2.Human Message***

##### **Human Message means Message provided by the user as input to the LLM**

```python
from langchain_core.messages import SystemMessage,HumanMessage

messages = [
    SystemMessage(content = "You are an AI Expert. please answer the following question in a Professional Way.In a Single Paragraph"),
    HumanMessage(content = "What is LangChain?")

]

response = llm.invoke(messages)
print(response.content)
```

#### ***3. AI Message***
##### **AI Message which is generated by AI**

```python
print(response.content)
```

##### ***We have an issue with Memory, Let's control it in a basic way using extra variable***

##### ***This may lead to the Context Window***

```python
chat_history = [
    SystemMessage(content = "You are an Helpful AI Assistant.")
]

while True:
    user_input = input("You:")

    if user_input.lower()=="exit":
        break
    print("User:",user_input)
    chat_history.append(HumanMessage(content = user_input))
    response = llm.invoke(chat_history)
    chat_history.append(response)

    print("Bot:",response.content)
```
