#### **Chat Prompt Template**

##### *Chat Prompt Template means it's reusable template for creating list of messages*

### ***Set and Load the API Key***

```python
import os
from dotenv import load_dotenv
load_dotenv()

if os.environ['GROQ_API_KEY']:
    print("API Key is Set")
```

### ***Create a LLM Model Using Groq Model***

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model = "openai/gpt-oss-120b"
)

llm
```

#### **Design a Prompt using Chat Prompt Template**

```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    ("system","You are an {domain} expert."),
    ("human","Explain to me in simple terms about {topic}.")
])

prompt = prompt_template.invoke({
    "domain":"AI Engineer",
    "topic":"Fine-Tuning"
})
prompt
```

```python
response = llm.invoke(prompt)
print(response.content)
```

```python
prompt = prompt_template.invoke({
    "domain":"AI",
    "topic":"RAG"
})
response = llm.invoke(prompt)
print(response.content)
```
