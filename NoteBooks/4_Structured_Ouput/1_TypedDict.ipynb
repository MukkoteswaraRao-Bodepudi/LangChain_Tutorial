#### ***Structured Output Means ask the LLM to generate response in a specific structure instead of free returning text***

```python
## Load API Key
import os
from dotenv import load_dotenv
load_dotenv()

if os.environ['GROQ_API_KEY']:
    print("API Key is Set")
```

```python
### Create LLM
from langchain_groq import ChatGroq

llm = ChatGroq(
    model = "openai/gpt-oss-120b"
)

llm
```

```python
### Design a Prompt
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
from langchain_core.messages import SystemMessage,HumanMessage,AIMessage

prompt_template = ChatPromptTemplate.from_messages([
    ("system","You are an helpful Assistant"),
    ("human","{question}")
])
```

```python
### Create a Schema to generate a structured response
from typing import TypedDict,List
class Movie_Review(TypedDict):
    movie_summary:str
    cast:List
    rating:float
```

```python
### Integrate the schema with 

structured_output = llm.with_structured_output(Movie_Review)

response = structured_output.invoke("Can I Know about the RRR Movie?")
response
```
