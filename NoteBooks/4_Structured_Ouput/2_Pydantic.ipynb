```python
###Load API Key
import os
from dotenv import load_dotenv
load_dotenv()

if os.environ['GROQ_API_KEY']:
    print("API Key is Set")
```

```python
### Create LLM
from langchain_groq import ChatGroq

llm = ChatGroq(
    model = "openai/gpt-oss-120b"
)

llm
```

```python
### create a Pydantic Schema
from pydantic import BaseModel,Field

class ProductReview(BaseModel):
    sentiment:str = Field(description = "Overall Sentiment: Positve, Negative, Neutral")
    summary:str = Field(description = "Summary of the Product")
    rating:int = Field(description = "Rating of Product from 1 to 5")
```

```python
#### Integrates the structured output with llm model
structured_output = llm.with_structured_output(ProductReview)

response = structured_output.invoke(
    """ 
    The phone has an excellent camera and battery.
    Performance is very good, but the device is
    slightly expensive.
    I would give it 4 stars.
    """
)
print(response)
```
