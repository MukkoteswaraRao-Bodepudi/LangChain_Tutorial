#### ***Directory Loader***

##### ***Directory Loader is used when you have multiple files inside a folder, and want to load them together instead of creating a seperate loader for every file***

```python
from langchain_community.document_loaders import TextLoader
from langchain_community.document_loaders import DirectoryLoader

loader = DirectoryLoader(
    "Docs",
    glob = "*.txt",
    loader_cls=TextLoader)

documents = loader.load()

documents
```
