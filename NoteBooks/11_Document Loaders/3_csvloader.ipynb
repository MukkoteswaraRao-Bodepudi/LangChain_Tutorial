#### **CSV Loader**
##### ***CSV Loader is a Document Loader is used to load data from CSV File and Convert the rows into a LangChain Document objects.***

```python
from langchain_community.document_loaders import CSVLoader
import os
### Path Validation
path = "Docs/medical_charges.csv"

if os.path.exists(path):
    print("Path is Exist")
else:
    print("Path is not Exist")
```

```python
loader = CSVLoader(path)

documents = loader.load()

documents
```

```python
print("Number of Rows in csv File:",len(documents))
```
