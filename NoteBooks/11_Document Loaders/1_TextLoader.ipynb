### **Document Loader**

#### ***Document Loaders is a LangChain Component is used to load the data from source and converts into a LangChain Document.***


##### **Document has two important parts**

##### ***1.page_content: It contains actual text of Document.***

##### ***2.metadata: Information about where content came from***

#### ***1.TextLoader***
##### **Text Loader is a Document Loader is used to load text files such as .txt file and convert it into Document Objects.**

```python
from langchain_community.document_loaders import TextLoader
import os
### Path Validation
path = "Docs/langsmith.txt"

if os.path.exists(path):
    print("Path is Exist")
else:
    print("Path is not Exist")
```

```python
loader = TextLoader(path)

documents = loader.load()

documents
```
