#### ***WebBaseLoader***

##### **WebBaseLoader is a document loader is used to load the web page content from webpages and converts webpage into Document Objecs.**

##### **You can load multiple urls**

```python
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader("https://en.wikipedia.org/wiki/Harley-Davidson")

documents = loader.load()
documents
```
