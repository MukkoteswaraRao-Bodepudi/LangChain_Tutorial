#### ***Structure Output Parser generate the response from LLM in a specified Format***

```python
###Load API Key
import os
from dotenv import load_dotenv
load_dotenv()

if os.environ['GROQ_API_KEY']:
    print("API Key is Set")
```

```python
### Create LLM
from langchain_groq import ChatGroq

llm = ChatGroq(
    model = "openai/gpt-oss-120b"
)

llm
```

```python
from pydantic import BaseModel,Field

class Candidate(BaseModel):

    name:str = Field(description="The Person's full name")
    age:int = Field(description="gt = 18 ls = 100 Person's age")
    place:str = Field(description="Place of the Person")
```

```python
from langchain_core.output_parsers import PydanticOutputParser

output_parser = PydanticOutputParser(pydantic_object=Candidate)

format_instructions = output_parser.get_format_instructions()
```

```python
### Design a Prompt
from langchain_core.prompts.prompt import PromptTemplate

prompt_template = PromptTemplate.from_template("""
    Provide the frictional name, age and {place} of Person. Make Sure age must be greater than 18 and less than 100.
    Generate the Response based on {format_instructions}
""")
```

```python
### Chain
chain = prompt_template | llm | output_parser

response = chain.invoke({
    "place":"Germany",
    "format_instructions":output_parser.get_format_instructions()
})
print(response)
```
